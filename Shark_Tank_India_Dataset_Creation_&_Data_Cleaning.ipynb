```python
# ============================================================
# STEP 4 — SCRAPE HTML TABLES USING pd.read_html()
# ============================================================

import pandas as pd
import requests

# Source website
url = "https://en.wikipedia.org/wiki/Shark_Tank_India_season_1"

# Browser-like headers to avoid HTTP 403
headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/153.0.0.0 Safari/537.36"
    )
}

# Fetch webpage
response = requests.get(url, headers=headers, timeout=30)

print("Status Code:", response.status_code)

# Check whether webpage was successfully fetched
response.raise_for_status()

# Scrape all HTML tables using pandas
tables = pd.read_html(response.text)

print("Number of tables scraped:", len(tables))

# Display basic information about scraped tables
for i, table in enumerate(tables):
    print(f"\nTable {i}: {table.shape}")
    display(table.head(3))
```

## 5. Select the Required Dataset

The scraped webpage contains multiple HTML tables.

For this project, the main dataset is the **pitch and investment table**, which contains information about businesses presented on Shark Tank India, including their episode number, pitch number, brand, business idea, city, original funding ask, deal information, and shark investments.

The required table is selected from the scraped HTML tables for further data cleaning, transformation, and optimization.


```python
# ============================================================
# STEP 5 — SELECT THE REQUIRED DATASET
# ============================================================

# The main pitch and investment table is Table 2
shark_tank_df = tables[2].copy()

# Display dataset dimensions
print("Dataset Shape:", shark_tank_df.shape)

# Display column names
print("\nColumn Names:")
print(shark_tank_df.columns.tolist())

# Display first 5 rows
print("\nFirst 5 Rows:")
display(shark_tank_df.head())
```

## 6. Data Cleaning — Column Names and Repeated Headers

The scraped dataset contains multi-level column headers because the original webpage groups the shark investment columns under a common "Invested by" heading.

For better usability and analysis, the following cleaning operations are performed:

- Flatten the multi-level column headers.
- Retain meaningful shark names as individual columns.
- Remove unnecessary duplicate header information.
- Remove the repeated header row present inside the dataset.
- Reset the DataFrame index.
- Standardize column names for easier analysis.

```python
# ============================================================
# STEP 6 — CLEAN COLUMN NAMES AND REMOVE REPEATED HEADERS
# ============================================================

# Create a copy so the scraped dataset remains unchanged
clean_df = shark_tank_df.copy()

# ------------------------------------------------------------
# 1. Flatten MultiIndex columns
# ------------------------------------------------------------

if isinstance(clean_df.columns, pd.MultiIndex):

    cleaned_columns = []

    for column in clean_df.columns:

        # Remove empty / NaN values
        parts = [
            str(part).strip()
            for part in column
            if pd.notna(part) and str(part).strip() != ""
        ]

        # Remove duplicate consecutive values
        unique_parts = []
        for part in parts:
            if part not in unique_parts:
                unique_parts.append(part)

        # For grouped columns such as "Invested by -> Aman Gupta",
        # keep the meaningful lower-level column name.
        if len(unique_parts) > 1:
            column_name = unique_parts[-1]
        elif len(unique_parts) == 1:
            column_name = unique_parts[0]
        else:
            column_name = "Unknown"

        cleaned_columns.append(column_name)

    clean_df.columns = cleaned_columns

else:
    # If columns are already single-level
    clean_df.columns = [
        str(column).strip()
        for column in clean_df.columns
    ]


# ------------------------------------------------------------
# 2. Standardize column names
# ------------------------------------------------------------

clean_df.columns = (
    clean_df.columns
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)


# ------------------------------------------------------------
# 3. Remove repeated header row
# ------------------------------------------------------------

# The scraped table contains a repeated header row inside
# the actual dataset. Remove it using the Brand column.
if "Brand" in clean_df.columns:

    clean_df = clean_df[
        clean_df["Brand"].astype(str).str.strip().ne("Brand")
    ]


# ------------------------------------------------------------
# 4. Reset index
# ------------------------------------------------------------

clean_df = clean_df.reset_index(drop=True)


# ------------------------------------------------------------
# 5. Convert Episode and Pitch numbers to numeric
# ------------------------------------------------------------

for column in ["Ep. No.", "Pitch No."]:
    if column in clean_df.columns:
        clean_df[column] = pd.to_numeric(
            clean_df[column],
            errors="coerce"
        )


# ------------------------------------------------------------
# 6. Display cleaned dataset information
# ------------------------------------------------------------

print("Cleaned Dataset Shape:", clean_df.shape)

print("\nCleaned Column Names:")
print(clean_df.columns.tolist())

print("\nFirst 5 Rows:")
display(clean_df.head())

print("\nData Types:")
display(clean_df.dtypes)
```

## 7. Missing Values Analysis and Cleaning

Missing values are analyzed across all columns to understand the completeness of the scraped dataset.

The following steps are performed:

- Count missing values in each column.
- Calculate the percentage of missing values.
- Identify columns containing missing data.
- Determine an appropriate treatment based on the nature of the data.
- Replace missing categorical values with a consistent label where appropriate.

```python
# ============================================================
# STEP 7 — MISSING VALUES ANALYSIS AND CLEANING
# ============================================================

# Create a missing-value summary
missing_summary = pd.DataFrame({
    "Missing Values": clean_df.isna().sum(),
    "Missing Percentage": (
        clean_df.isna().mean() * 100
    ).round(2)
})

# Keep only columns containing missing values
missing_summary = (
    missing_summary[
        missing_summary["Missing Values"] > 0
    ]
    .sort_values("Missing Values", ascending=False)
)

print("Missing Value Summary:")
display(missing_summary)

# ------------------------------------------------------------
# Replace missing categorical/text values
# ------------------------------------------------------------

# Shark investment columns
shark_columns = [
    "Anupam Mittal",
    "Aman Gupta",
    "Ashneer Grover",
    "Namita Thapar",
    "Peyush Bansal",
    "Vineeta Singh",
    "Ghazal Alagh"
]

# Replace missing values in shark investment columns
# with "No Investment Recorded"
for column in shark_columns:
    if column in clean_df.columns:
        clean_df[column] = clean_df[column].fillna(
            "No Investment Recorded"
        )

# Replace missing values in general categorical columns
categorical_columns = ["Brand", "Idea", "City", "Deal"]

for column in categorical_columns:
    if column in clean_df.columns:
        clean_df[column] = clean_df[column].fillna(
            "Not Available"
        )

# ------------------------------------------------------------
# Verify missing values after cleaning
# ------------------------------------------------------------

remaining_missing = clean_df.isna().sum()

print("\nRemaining Missing Values:")
display(
    remaining_missing[
        remaining_missing > 0
    ].to_frame("Missing Values")
)

print("\nDataset Shape After Missing Value Treatment:")
print(clean_df.shape)
```

## 8. Duplicate Check and Data Type Optimization

The cleaned dataset is checked for duplicate records to avoid repeated observations.

The data types of important columns are also optimized to improve consistency and usability for further analysis.

The following operations are performed:

- Identify duplicate rows.
- Remove exact duplicate records, if present.
- Convert episode and pitch numbers to integer type.
- Convert appropriate text columns to string type.
- Verify the final data types and dataset structure.

```python
# ============================================================
# STEP 8 — DUPLICATE CHECK AND DATA TYPE OPTIMIZATION
# ============================================================

# ------------------------------------------------------------
# 1. Check for duplicate rows
# ------------------------------------------------------------

duplicate_count = clean_df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_count)

# Remove exact duplicate rows, if any
if duplicate_count > 0:
    clean_df = clean_df.drop_duplicates().reset_index(drop=True)

print("Dataset Shape After Duplicate Removal:", clean_df.shape)


# ------------------------------------------------------------
# 2. Optimize numeric columns
# ------------------------------------------------------------

for column in ["Ep. No.", "Pitch No."]:
    if column in clean_df.columns:
        clean_df[column] = pd.to_numeric(
            clean_df[column],
            errors="coerce"
        ).astype("Int64")


# ------------------------------------------------------------
# 3. Standardize text columns
# ------------------------------------------------------------

text_columns = [
    "Brand",
    "Idea",
    "City",
    "Original ask",
    "Deal",
    "Anupam Mittal",
    "Aman Gupta",
    "Ashneer Grover",
    "Namita Thapar",
    "Peyush Bansal",
    "Vineeta Singh",
    "Ghazal Alagh"
]

for column in text_columns:
    if column in clean_df.columns:
        clean_df[column] = clean_df[column].astype("string").str.strip()


# ------------------------------------------------------------
# 4. Verify data types
# ------------------------------------------------------------

print("\nFinal Data Types:")
display(clean_df.dtypes.to_frame("Data Type"))


# ------------------------------------------------------------
# 5. Final dataset information
# ------------------------------------------------------------

print("\nDataset Information:")
print("Rows    :", clean_df.shape[0])
print("Columns :", clean_df.shape[1])

print("\nDuplicate Rows Remaining:", clean_df.duplicated().sum())
```

```python
# ============================================================
# STEP 9 — DATA VALIDATION AND QUALITY CHECK
# ============================================================

print("=" * 60)
print("FINAL DATA VALIDATION")
print("=" * 60)

# ------------------------------------------------------------
# 1. Dataset dimensions
# ------------------------------------------------------------

print("\n1. Dataset Dimensions")
print("Rows    :", clean_df.shape[0])
print("Columns :", clean_df.shape[1])


# ------------------------------------------------------------
# 2. Missing value check
# ------------------------------------------------------------

total_missing = clean_df.isna().sum().sum()

print("\n2. Missing Value Check")
print("Total Missing Values:", total_missing)

if total_missing == 0:
    print("Status: PASS — No missing values found.")
else:
    print("Status: REVIEW — Missing values are present.")


# ------------------------------------------------------------
# 3. Duplicate check
# ------------------------------------------------------------

duplicate_count = clean_df.duplicated().sum()

print("\n3. Duplicate Check")
print("Duplicate Rows:", duplicate_count)

if duplicate_count == 0:
    print("Status: PASS — No duplicate rows found.")
else:
    print("Status: REVIEW — Duplicate rows are present.")


# ------------------------------------------------------------
# 4. Column name uniqueness
# ------------------------------------------------------------

unique_columns = clean_df.columns.is_unique

print("\n4. Column Name Check")
print("All Column Names Unique:", unique_columns)

if unique_columns:
    print("Status: PASS")
else:
    print("Status: REVIEW — Duplicate column names found.")


# ------------------------------------------------------------
# 5. Important field checks
# ------------------------------------------------------------

print("\n5. Key Field Check")

key_columns = ["Ep. No.", "Pitch No.", "Brand", "Idea", "City"]

for column in key_columns:
    if column in clean_df.columns:
        print(f"{column}: Available")


# ------------------------------------------------------------
# 6. Data types
# ------------------------------------------------------------

print("\n6. Data Types")
display(clean_df.dtypes.to_frame("Data Type"))


# ------------------------------------------------------------
# 7. Final preview
# ------------------------------------------------------------

print("\n7. Final Dataset Preview")
display(clean_df.head(10))


# ------------------------------------------------------------
# 8. Validation summary
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("VALIDATION SUMMARY")
print("=" * 60)

print("Rows:", clean_df.shape[0])
print("Columns:", clean_df.shape[1])
print("Missing Values:", total_missing)
print("Duplicate Rows:", duplicate_count)
print("Unique Column Names:", unique_columns)
```
